```python
ord_data.query("Segment=='Corporate'").groupby("Region").agg({"Sales":"sum","Profit":"mean","Quantity":"sum"})
```

```python
ord_data.query("Segment=='Corporate'").groupby("Region").agg({"Sales":"sum","Profit":"mean","Quantity":"sum"}).query("Quantity> 2100")
```

```python
ord_data.query("Segment=='Corporate'").groupby("Region").agg({"Sales":"sum","Profit":"mean","Quantity":"sum"}).query("Quantity> 2100").rename(columns= {"Sales":"Total_Sales"})
```

```python
ord_data.query("Segment=='Corporate'") \
     .groupby("Region") \
    .agg({"Sales":"sum","Profit":"mean","Quantity":"sum"}) \
    .query("Quantity> 2100") \
    .rename(columns= {"Sales":"Total_Sales"})
```

```python
ord_data.query("Segment=='Corporate'") \
    .groupby("Region") \
    .agg({"Sales":"sum","Profit":"mean","Quantity":"sum"}) \
    .query("Quantity>2100") \
    .rename(columns={"Sales": "Total_Sales","Profit":"Avg_Profit","Quantity":"Total_Quantity"}) \
    .sort_values("Total_Quantity",ascending=False) \
    .reset_index() # this will bring tht column header to normal position
```
