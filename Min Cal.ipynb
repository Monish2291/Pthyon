```python
def add (a,b):
    c = (a + b)
    print ("The addition of two nubers:" ,c)
```

```python
def sub (a,b):
    c = (a - b)
    print("The substraction of two numbers:", c)
```

```python
def mul (a,b):
    c = (a * b)
    print("The multipliction of two numbers:",c)
```

```python
def div(a,b):
    c= (a/b)
    print("The division of two numbers:",c)
```

```python
print("Please enter the first integer:")
a = int(input())

print("Please enter the second integer:")
b = int(input())

print("Please enter which operation you want: add/sub/mul/div")
o = input()

if o == "add":
    add(a, b)
elif o == "sub":
    sub(a, b)
elif o == "mul":
    mul(a, b)
elif o == "div":
    div(a, b)
else:
    print("Please select a valid operation: add/sub/mul/div")
```
